**Products** Table creating and inserting

```python
products_data = [
    ("P01", "Car Battery", "Car Parts"),
    ("P02", "Brake Pads", "Car Parts"),
    ("P03", "Engine Oil", "Car Care"),
    ("P04", "Bike Helmet", "Cycling"),
    ("P05", "Bike Light", "Cycling"),
    ("P06", "Roof Box", "Travel"),
    ("P07", "Wiper Blades", "Car Parts"),
    ("P08", "Car Shampoo", "Car Care")
]

products_schema = StructType([
    StructField("product_id", StringType(), False),
    StructField("product_name", StringType(), False),
    StructField("category", StringType(), False)
])

products_df = spark.createDataFrame(products_data, products_schema)
```

**Customers** Table creating and inserting

```python
customers_data = [
    ("  c001", "Ravi Kumar", " BS1 1AA"),
    ("c002", "John Smith", "M1 2AB  "),
    ("C003", "Sarah Jones", None),
    (" C003", "Sarah Jones", "B1 1AA "),
    ("c004", "David Lee", "L1 5XY"),
    ("C005", "Mike Brown", None),

]

customers_schema = StructType([
    StructField("customer_id", StringType(), False),
    StructField("customer_name", StringType(), False),
    StructField("postcode", StringType(), True)
])

customers_df = spark.createDataFrame(customers_data, customers_schema)
```

Orders Table creating and inserting

```python
orders_data = [
    ("O1001", "C001", "2026-06-01", "Completed", "2026-06-01 10:00"),
    ("O1001", "C001", "2026-06-01", "Completed", "2026-06-01 12:00"),
    ("O1002", "c002", "2026-06-15", "Cancelled", "2026-06-15 11:00"),
    ("O1002", "c002", "2026-06-15", "Completed", "2026-06-15 15:00"),
    ("O1003", "C003", "2026-07-01", "Completed", "2026-07-01 09:00"),
    ("O1004", "C003", "2027-01-01", "Completed", "2026-07-02 10:00"),
    ("O1005", "C004", "2026-07-05", "Completed", "2026-07-05 10:00"),
    ("O1006", "C005", "2026-07-10", "Cancelled", "2026-07-10 10:00")
]

orders_schema = StructType([
    StructField("order_id", StringType(), False),
    StructField("customer_id", StringType(), False),
    StructField("order_date", StringType(), False),
    StructField("order_status", StringType(), False),
    StructField("updated_at", StringType(), False)
])

orders_df = spark.createDataFrame(orders_data, orders_schema)
```

```python
order_items_data = [
    ("O1001", "P01", 2, 50.0),
    ("O1001", "P02", 1, 100.0),
    ("O1002", "P03", 2, 75.0),
    ("O1003", "P01", 3, 50.0),
    ("O1003", "P04", -2, 500.0),
    ("O1004", "P05", 2, 200.0),
    ("O1005", "P06", 1, -100.0),
    ("O1005", "P07", 2, 40.0),
    ("O1006", "P08", 3, 25.0)
]

order_items_schema = StructType([
    StructField("order_id", StringType(), False),
    StructField("product_id", StringType(), False),
    StructField("quantity", IntegerType(), False),
    StructField("unit_price", DoubleType(), False)
])

order_items_df = spark.createDataFrame(
    order_items_data,
    order_items_schema
)
```

## customer data cleaning from customer_id and postcode **bold text**

```python
customer_df_1=( customers_df.withColumn("customer_id",upper(trim(col("customer_id"))))
.withColumn("postcode",trim(regexp_replace(col("postcode")," ","")))
)
```

```python
# df_1 = customer_df_1.select(col("customer_id"),col("postcode"),length(col("customer_id")))
```

**Duplicate customers**

```python
w_customer= w.partitionBy(col("customer_id")).orderBy(col("postcode").desc())
customer_df_2 = (customer_df_1.withColumn("rn",row_number().over(w_customer))
                 .filter(col("rn")==1).drop(col("rn"))
)
# customer_df_2.show()
```

first filter latest data and then only consider completed one

```python
w_order = w.partitionBy(col("order_id")).orderBy(col("updated_at").desc())
order_df1= orders_df.withColumn("rn",row_number().over(w_order)).filter(col("rn")==1).drop(col("rn"))
order_df2 = order_df1.filter(col("order_status")=="Completed")
order_df1.show()
```
